```python
import cv2
import pandas as pd
from ultralytics import YOLO
import os

MEDIA_FOLDER = "/content/drive/MyDrive/traffic_project/data"

CONF_THRESHOLD = 0.4
MINI_TRUCK_MAX_AREA = 0.15
MINI_TRUCK_MAX_AR = 1.8

CAR_ID = 2
MOTORCYCLE_ID = 3
TRUCK_ID = 7

SUPPORTED_EXTENSIONS = (".mp4", ".avi", ".mov", ".mkv", ".jpg", ".jpeg", ".png")

model = YOLO("yolov8n.pt")
```

```python
def classify_vehicle(class_id, bbox_area, frame_area, aspect_ratio):
    if class_id == MOTORCYCLE_ID:
        return "two_wheeler"
    if class_id == CAR_ID:
        return "car"
    if class_id == TRUCK_ID:
        if (bbox_area / frame_area) < MINI_TRUCK_MAX_AREA and aspect_ratio < MINI_TRUCK_MAX_AR:
            return "mini_truck"
        return "truck"
    return None
```

```python
def process_image(image_path):
    frame = cv2.imread(image_path)
    if frame is None:
        return None

    h, w = frame.shape[:2]
    frame_area = h * w

    counts = {"car": 0, "truck": 0, "mini_truck": 0, "two_wheeler": 0}

    results = model(frame, conf=CONF_THRESHOLD, verbose=False)

    for r in results:
        if r.boxes is None:
            continue

        for box in r.boxes:
            class_id = int(box.cls[0])
            x1, y1, x2, y2 = box.xyxy[0]

            width = x2 - x1
            height = y2 - y1
            vehicle = classify_vehicle(
                class_id,
                width * height,
                frame_area,
                width / height
            )

            if vehicle:
                counts[vehicle] += 1

    return counts
```

```python
def process_video(video_path):
    cap = cv2.VideoCapture(video_path)
    tracked = {}

    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break

        h, w = frame.shape[:2]
        frame_area = h * w

        results = model.track(
            frame,
            conf=CONF_THRESHOLD,
            persist=True,
            tracker="bytetrack.yaml",
            verbose=False
        )

        for r in results:
            if r.boxes is None or r.boxes.id is None:
                continue

            for box, track_id in zip(r.boxes, r.boxes.id):
                track_id = int(track_id)
                if track_id in tracked:
                    continue

                class_id = int(box.cls[0])
                x1, y1, x2, y2 = box.xyxy[0]

                width = x2 - x1
                height = y2 - y1
                vehicle = classify_vehicle(
                    class_id,
                    width * height,
                    frame_area,
                    width / height
                )

                if vehicle:
                    tracked[track_id] = vehicle

    cap.release()

    counts = {"car": 0, "truck": 0, "mini_truck": 0, "two_wheeler": 0}
    for v in tracked.values():
        counts[v] += 1

    return counts
```

```python
results = []
file_id = 1

for file in sorted(os.listdir(MEDIA_FOLDER)):
    if not file.lower().endswith(SUPPORTED_EXTENSIONS):
        continue

    path = os.path.join(MEDIA_FOLDER, file)
    print(f"▶ Processing {file}")

    if file.lower().endswith((".jpg", ".jpeg", ".png")):
        counts = process_image(path)
    else:
        counts = process_video(path)

    if counts is None:
        continue

    counts["id"] = file_id
    counts["filename"] = file
    results.append(counts)
    file_id += 1

output_df = pd.DataFrame(results)
output_df = output_df[["id", "filename", "car", "truck", "mini_truck", "two_wheeler"]]
output_df.to_csv("output.csv", index=False)

output_df
```
